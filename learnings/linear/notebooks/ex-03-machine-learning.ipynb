## SETUP
We simulate 150 samples. The true rule is `y = 4 + 2.5*x1 - 1.8*x2 + noise`. The columns `x3, x4, x5` are pure noise and carry no information.

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
rng = np.random.default_rng(42)
n = 150
X = rng.normal(size=(n, 5))
y = 4 + 2.5 * X[:, 0] - 1.8 * X[:, 1] + rng.normal(scale=1.5, size=n)
df = pd.DataFrame(X, columns=["x1", "x2", "x3", "x4", "x5"]).round(3)
df.insert(0, "id", np.arange(1, n + 1))
df["y"] = y.round(3)
print(df.shape)
df.head()
```
